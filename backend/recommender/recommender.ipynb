```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import joblib
```

```python
df_cleaned = pd.read_csv(r'C:\Users\lenovo\Documents\BookSmart\data\livres_nettoyes.csv')
df_cleaned.columns
```

```python
# Extracting the 'Description' column for processing
descriptions = df_cleaned['Description']
```

```python
# 3. TF-IDF Vectorizer
vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')
tfidf_matrix = vectorizer.fit_transform(descriptions)
```

```python
# 4. Calcul de la matrice de similarité cosinus
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
```

```python
df_cleaned.loc[10]
```

```python
similarities=cosine_sim[10]
```

```python
# Trier par ordre décroissant de similarité
sorted_indices = np.argsort(similarities)[::-1]
```

```python
# Retirer l'index 10 (le livre lui-même)
sorted_indices = sorted_indices[sorted_indices != 10]
```

```python
# Afficher les indices et scores des 10 livres les plus similaires
for i in sorted_indices[:10]:
    print(f"Livre {i} avec similarité {similarities[i]:.3f}")
```
